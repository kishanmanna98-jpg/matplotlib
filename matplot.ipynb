```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("line plot ex")
plt.xlim(1,10)
plt.ylim(5,15)
plt.savefig("matplot")
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as pl
x=[101,102,103,104]
y=[100,72,38,40]
x1=[105,106,107,108]
y1=[99,65,58,29]
pl.xlabel('x axis')
pl.ylabel('y axis')
pl.title('mark')
pl.plot(x,y,label='line 1',color='red',linestyle='dotted',linewidth=3,marker='D',markerfacecolor='pink',markersize=20)
pl.plot(x1,y1,label='line 2',color='red',linestyle='dotted',linewidth=3,marker='*',markerfacecolor='pink',markersize=20)
pl.savefig("matplot")
pl.legend()
pl.plot(x,y)
pl.show()
```

```python
import matplotlib.pyplot as pl
x=[10,20,30,40,50]
y=[18,25,32,40,56]
x1=[44,52,62,58,25]
y1=[52,21,62,35,44]
pl.xlabel('a axis')
pl.ylabel('b axis')
pl.plot(x1,y1)
pl.plot(x,y,color='pink',linestyle='dashdot',linewidth=4,marker='*',markerfacecolor='purple',markersize=30)
pl.legend()
pl.show()
```

```python
import matplotlib.pyplot as pl
stu=['A1','B2','C3','D4']
marks=[100,72,38,70]
pl.bar(stu,marks,color='blue',width=0.8,edgecolor='purple',label='student marks')
pl.legend()
pl.title('student mark')
pl.xlabel('student')
pl.ylabel('marks')       
pl.show()       
```

```python
import pandas as p
a=p.read_csv('iris.csv')
x=a["sepal length"][0:5]
y=a["sepal width"][0:5]
pl.title('iris')
pl.xlabel('x')
pl.ylabel('y')
pl.bar(x,y,color='blue',width=0.8,edgecolor='purple',label='iris')
pl.legend()
pl.show()
pl.scatter(x,y,color='red',marker='*',s=102,label='point')
pl.legend()
```
